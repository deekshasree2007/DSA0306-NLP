```python
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Input sentence
sentence = input("Enter a sentence: ")

# Process the sentence
doc = nlp(sentence)

print("\nNoun Phrases and Their Meanings:")
print("--------------------------------")

for chunk in doc.noun_chunks:
    print("Noun Phrase :", chunk.text)

    # Display the main noun (root)
    print("Main Noun   :", chunk.root.text)

    # Display simple meaning
    print("Meaning     :", "Represents", chunk.root.text)
    print()
```

```python
import spacy

# Load English language model
nlp = spacy.load("en_core_web_sm")

# Input text
text = input("Enter a sentence: ")

# Process the text
doc = nlp(text)

# Store previously mentioned nouns
nouns = []

print("\nReference Resolution:")
print("---------------------")

for token in doc:

    # Store nouns and proper nouns
    if token.pos_ in ["NOUN", "PROPN"]:
        nouns.append(token.text)

    # Check for pronouns
    if token.pos_ == "PRON":
        if nouns:
            print(token.text, "->", nouns[-1])
        else:
            print(token.text, "-> No reference found")
```
